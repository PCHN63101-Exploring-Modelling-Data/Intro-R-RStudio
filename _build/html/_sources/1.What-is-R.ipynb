
R is a programming language and, like most programming languages, it is especially useful for certain kinds of tasks. R was designed for working with data: conducting statistical analyses, transforming datasets, and creating visualisations. This makes it particularly useful to scientists and other researchers, but it is also used in industry, government, and journalism. One of the advantages of such a large user base is that it is very likely that someone will already have solved a problem similar to the one you are trying to solve.

A particular benefit of R is that it is open source and freely available. Aside from saving money, this matters in science because it reduces barriers to reproducing analyses. If you publish your R analysis scripts and your data alongside a scientific paper, other researchers can rerun your analysis without needing to purchase a software licence. This is one reason R has become so important in modern research.

R evolved from an earlier programming language called S. As R grew steadily in popularity during the 1990s and early 2000s it became one of the main programming languages used for statistical analysis in universities and research settings. Part of that growth came from the fact that users could contribute packages and tools for the kinds of problems they regularly faced, be they in psychology, genetics, economics, or any number of other fields.



## Why not use SPSS?

Why, then, are so many psychology undergraduates taught SPSS? One reason is that SPSS can be easier to approach initially. Its menus and dialogue boxes allow you to select an ANOVA, for example, and drag variables into the relevant fields without writing any code. Programming is a technical skill, and learning any programming language can feel intimidating at first.

However, R offers considerably greater flexibility. In SPSS, you are largely limited to the procedures and options provided by the software. As a programming language R can do much more. It can import and clean data, scrape information from the web, perform almost any statistical analysis, generate publication-ready visualisations, and produce elegantly formatted reports. If a particular function is not already available, it may be possible to extend R or write the required functionality yourself, but it's rare that someone hasn't already done this for you.

Producing plots in R takes practice, but it is also one of the most enjoyable aspects of working with the language. Deciding how to visualise your data and analyses is an important and creative part of doing science. If you can think of a useful way to visualise your data, there is a good chance that R can produce it.

SPSS remains a useful tool, particularly for beginners and for researchers who prefer a graphical interface. Learning R, however, provides a more flexible and reproducible approach while developing transferable skills in managing, analysing, and reporting data.
