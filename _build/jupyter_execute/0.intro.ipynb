# Week 2: Introduction to R
By the end of this unit you should be able to take a data file, inspect it, clean it, summarise it, visualise it, conduct the appropriate statistical analysis, and then turn that whole workflow into a reproducible report. You can achieve all of that using one programming language: R.

The first five weeks of this unit will focus on getting you up to speed in R. No previous knowledge of R is assumed and you do not need to worry if you have never programmed before. We will start with the basics and gradually add more complex topics each week. There will be plenty of opportunities to put this learning into practice during the in-person sessions.

Once you understand how R works, the last five weeks of the unit will show you how to build statistical models in R and answer your research questions. The focus in the second half will be on statistics: choosing the appropriate way to model your data and interpreting the output.

This week, you’ll be introduced to key concepts in the programming language R and to RStudio, the software we use to work with R. At this stage you are not expected to memorise everything or be able to write code independently. The aim this week is to become familiar with the environment and some of the core ideas. 

As you work through this week's content there is no expectation that you run any of the code examples you see. But if you would like to experiment, head over to the sandbox on posit cloud here: https://posit.cloud/spaces/797313/content/12996416


Don't worry if it doesn't work or you get stuck. We will go through everything in the in-person session.
