# Vectors
One of the most important kinds of object in R is the vector. A vector is a collection of values stored together in a single object. All the values in a vector must be of the same basic type, such as numeric, character, or logical.

## Creating vectors

To create a vector we can use the combine function, written simply as `c()`.

For example: 

```r
scores <- c(55, 65, 60, 70, 65)
```

This creates a single vector object called `scores` which contains multiple values. Vectors don't have to contain numbers: 

```r
bandMembers <- c("Thom", "Colin", "Ed", "Phil","Jonny")
```







## Vector types

The most common vector types you will encounter are:

* numeric: numbers like we have in `scores`
* character: text such as `"Thom" `or `"hello"`
* logical: values that are either `TRUE` or `FALSE`

You can check what kind of vector you have using `class()`:

```r
class(scores)
```
```text
[1] "numeric"
```

You can also check how many elements a vector contains using `length()`

```r
length(band_members)

```
```text
[1] 5
```



````{warning}
Be careful: If you try and create a vector with different types of values in it, R will not complain. Instead R will coerce everything into the same type.

```r
georgeStats <- c('George',80,TRUE)
georgeStats 
```
```text
[1] "George" "80"     "TRUE"
```
Note the output above shows all the values in quotation marks. This tells us that '80' and 'TRUE' are now just characters and have lost their original numeric and logical types. We can verify this using the `class()` function we saw earlier:

```r
class(georgeStats)
```
```text
[1] "character"
```
````

## Missing values

When you work with real world data it is often the case that values will be missing, perhaps one of your participants forgets to answer a survey question, or some piece of experiment equipment fails to record a reaction time. In vectors these missing values are often represented with `NA`. In R `NA` stands for 'not available' and you can use it within any type of vector.

## Vectorised operations

Vectors are very useful because you can easily perform operations on all their elements at once. For example, suppose we wanted to add 2 to all of the values in the `scores` vector we created above:

```r
scores + 2

```
gives the following output:

```text
[1] 57, 67, 62, 72, 67
```

The output is the result of adding 2 to each element in the vector. You can do the same with other operations:

```r
scores * 2
```
```text
[1] 110 130 120 140 130
```
Each element was multiplied by 2.
```r
scores > 60
```
```text
[1] FALSE TRUE FALSE TRUE TRUE
```
The output tells us whether each element is greater than 60. 

This is an important idea in R: many operations work element by element across a whole vector.

```{tip}
If you are wondering why `scores > 60` didn't result in five `TRUE`s, it's because the code `scores * 2` was not assigned to an object. So the console displayed the result of the multiplications but the original `scores` vector remained unaffected. Had we wanted to change `scores` so that each value was multiplied by 2, we could have used `scores <- scores * 2`
```
