## Objects and assignment

On the previous page we saw that commands such as `2 + 2` typed directly into the console will return `4`. R carried out the calculation and displayed the result, but that result was not stored anywhere. If we want to keep a result and use it later, we need to store it in an object.

We do this using the assignment operator `<-`. This is typed using the less-than symbol (<) followed by a minus sign (-). The assignment operator tells R to take the value on the right-hand side and store it in the object named on the left-hand side. If the object named on the left-hand side already exists it will be overwritten. If it does not exist, R will create it. You can tell what objects already exist by looking in the environment pane.

In the example below we calculate two plus two and assign the result to a new object called `mySum` 

```r
mySum <- 2 + 2
```

The `mySum` object now exists and you will be able to see it in the environment pane. You can also inspect it by typing its name into the console, which would produce:

```text
[1] 4
```

This means the value has been stored and can now be reused. Suppose you wanted to add another 2 to your `mySum` object, the following command typed into the console would do it:

 ```r
 mySum <- mySum + 2
 ```

The value of `mySum` was already 4, so the code above effectively tells R to add 2 to 4 and assign the result to `mySum`, which overwrites the previous value of 4 with 6. 



### Types

The values stored in an object have different types which reflect whether they are numbers, text or some other kind of value. The code below uses the assign operator to create three objects with different types.

```r
score <- 55
firstName <- "Aisha"
passed <- score >= 50
```
`score` is a now a numeric object, `firstName` is a character object and `passed` is a logical value, which in this case is `TRUE` because the value of `score` (55) is indeed greater than 50.

R usually works out the type of object automatically based on what you assign. If you assign text, R creates a character object. If you assign a number, R creates a numeric object. If you assign the result of a comparison, R creates a logical object. R isn't always guaranteed to get this right, so you can check what type of object you are dealing with by using the `class()` function like this:

```r
class(passed)
```

```text
[1] "logical"

```



### Naming objects
When you name an object it must start with a letter and it can't contain spaces or special symbols like `@`,`!`,`+` or `-`.  Try to use names that are descriptive. So use `reactionTime` rather than `dv1`. If your name has more than one word you can use camel case or snake case. In camel case the second word starts with a capital letter: `reactionTime`. In snake case you place an underscore between the words: `reaction_time`. It's up to you what you use, but be consistent. You should also avoid calling objects names of functions in R like `mean` or `summary`. Dots `.` are permitted in names but are best avoided.
